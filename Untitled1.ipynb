```python

#1. Número Invertido
numero = int(input("Digite um número: "))

while numero > 0:
    print(numero % 10, end="")
    numero //= 10
```

```python
 # Contador de Dígitos
numero = int(input("Digite um número: "))

contador = 0

while numero > 0:
    contador += 1
    numero //= 10

print("Quantidade de dígitos:", contador)
```

```python
# Cofrinho
meta = 10000
total = 0
mes = 0

deposito = float(input("Digite o valor do depósito mensal: "))

while total < meta:
    total += deposito
    mes += 1

print("Meses necessários:", mes)
print("Total acumulado: R$", total)
print("Meta atingida no mês:", mes)
```

```python
#Caixa Registradora
subtotal = 0

while True:
    preco = float(input("Preço do produto (0 para finalizar): "))

    if preco == 0:
        break

    quantidade = int(input("Quantidade: "))
    subtotal += preco * quantidade

if subtotal >= 1000:
    desconto = 0.15
elif subtotal >= 500:
    desconto = 0.10
elif subtotal >= 200:
    desconto = 0.05
else:
    desconto = 0

valor_desconto = subtotal * desconto
valor_final = subtotal - valor_desconto

print("Subtotal: R$", subtotal)
print("Desconto: R$", valor_desconto)
print("Valor final: R$", valor_final)
```
