```python
# Questão 5

valor = int(input("Digite o valor do saque: R$ "))
nota = 100
while valor > 0:
    quantidade = valor // nota

    if quantidade > 0:
        print(f"{quantidade} notas de R${nota}")
    valor = valor % nota

    if nota == 100:
        nota = 50
    elif nota == 50:
        nota = 20
    elif nota == 20:
        nota = 10
    elif nota == 10:
        nota = 5
    elif nota == 5:
        nota = 2
    elif nota == 2:
        nota = 1
```
